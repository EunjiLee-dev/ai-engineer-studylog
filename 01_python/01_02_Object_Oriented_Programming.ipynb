### Exercise 1: AI Dataset Management System
Create a Dataset class that manages a collection of AI training samples.

**Requirements**
- Initialize a dataset with a name and a list of samples.
- Store dataset attributes such as name, samples, and created_at.
- Implement methods to add, remove, and retrieve samples.
- Implement a method to return the total number of samples.
- Implement a method to display dataset information.

Expected behavior: You should be able to create multiple datasets and manage them independently.

```python
from datetime import datetime

class Dataset:
    def __init__(self, name, samples=None):
        self.name = name
        self.samples = samples if samples is not None else []
        self.created_at = datetime.now()

    def add_sample(self, sample):
        self.samples.append(sample)
        print(f"Sample added successfully: {sample}")

    def remove_sample(self, sample):
        if sample in self.samples:
            self.samples.remove(sample)
            print(f"Sample removed successfully: {sample}")
        else:
            print(f"Sample not found: {sample}")

    def get_sample(self, index):
        if 0 <= index < len(self.samples):
            return self.samples[index]
        else:
            print("Invalid index")

    def get_total_number(self):
        return len(self.samples)

    def display_info(self):
        print("=== Dataset Information ===")
        print(f"Name: {self.name}")
        print(f"Created at: {self.created_at}")
        print(f"Total Samples: {self.get_total_number()}")

# samples
dataset1 = Dataset("Image Dataset", ["image1", "image2"])
dataset2 = Dataset("Text Dataset", ["text1", "text2", "text3"])

dataset1.add_sample("image3")
dataset1.remove_sample("image1")

print("\nRetrieved Sample:", dataset1.get_sample(0))
print("Total Samples:", dataset1.get_total_number())

dataset1.display_info()
dataset2.display_info()
```
