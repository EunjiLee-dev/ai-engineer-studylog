### Exercise 1: AI Training Dataset Validation Pipeline

You are working as an AI Engineer responsible for validating training datasets before model training.

Your team receives raw datasets from multiple sources. The data may contain missing values, duplicate IDs, invalid labels, and inconsistent text formatting.

**Requirements**
Create a function called validate_dataset() that validates the entire dataset.
Check whether required fields exist in every sample.
Identify samples containing:
Missing text
Empty text
Invalid labels
Duplicate IDs
Normalize text by removing unnecessary whitespace and converting it to lowercase.
Use sets to identify duplicate IDs and unique data sources.
Use list and dictionary comprehensions to create cleaned datasets and validation summaries.
Separate valid and invalid samples.
Return both datasets and a validation report.
Calculate the percentage of valid samples.

**Expected Output**
``` text
===== Dataset Validation Report =====

Total Samples: 8
Valid Samples: 4
Invalid Samples: 4
Duplicate IDs: [101]

Invalid Reasons:
- Missing Text: 1
- Empty Text: 1
- Invalid Label: 1
- Duplicate ID: 1

Valid Data Sources: ['A', 'B']
Validation Rate: 50.0%
```

````python
# given dataset
raw_dataset = [
    {"id": 101, "text": "  Machine Learning is great! ", "label": "positive", "source": "A"},
    {"id": 102, "text": "Deep Learning is powerful", "label": "positive", "source": "B"},
    {"id": 103, "text": "AI is difficult", "label": "negative", "source": "A"},
    {"id": 104, "text": "  ", "label": "positive", "source": "C"},
    {"id": 105, "text": "Python is useful", "label": "unknown", "source": "B"},
    {"id": 101, "text": "Machine Learning is great!", "label": "positive", "source": "A"},
    {"id": 107, "text": None, "label": "negative", "source": "C"},
    {"id": 108, "text": "Neural Networks are interesting", "label": "positive", "source": "A"},
]


def validate_dataset(dataset):
    required_fields = ["id", "text", "label", "source"]
    valid_labels = ["positive", "negative"]

    valid_samples = []
    invalid_samples = []
    checked_ids = set()
    duplicate_ids = set()

    error_cnt = {
        "missing_required_fields": 0,
        "missing_text": 0,
        "empty_text": 0,
        "invalid_label": 0,
        "duplicate_id": 0
    }

    for sample in dataset:
        errors = []
        normalized_sample = sample.copy()

        # check - required fields
        if not all(field in sample for field in required_fields):
            errors.append("missing_required_fields")
            error_cnt["missing_required_fields"] += 1

        # check - missing text
        if sample["text"] is None or "text" not in sample:
            errors.append("missing_text")
            error_cnt["missing_text"] += 1

        # check - empty text
        elif sample["text"].strip() == "":
            errors.append("empty_text")
            error_cnt["empty_text"] += 1

        # check - valid label
        if "label" in sample:
            if sample["label"] not in valid_labels:
                errors.append("invalid_label")
                error_cnt["invalid_label"] += 1

        # check - duplicate id
        if sample["id"] in checked_ids:
            duplicate_ids.add(sample["id"])
            errors.append("duplicate_id")
            error_cnt["duplicate_id"] += 1
        else:
            checked_ids.add(sample["id"])

        # normalize text
        if isinstance(sample["text"], str):
            normalized_sample["text"] = sample["text"].strip().lower()

        # valid?
        if errors:
            invalid_samples.append(normalized_sample)
        else: 
            valid_samples.append(normalized_sample)


    # valid source: at least one valid sample
    valid_sources = set(
        sample["source"] for sample in valid_samples
    )


    print("===== Dataset Validation Report =====\n")
    print(f"Total Samples: {len(dataset)}")
    print(f"Valid Samples: {len(valid_samples)}")
    print(f"Invalid Samples: {len(invalid_samples)}")
    print(f"Duplicate IDs: {sorted(duplicate_ids)}")
    print("\nInvalid Reasons:")
    print(f"- Missing Required Fields: {error_cnt['missing_required_fields']}")
    print(f"- Missing Text: {error_cnt['missing_text']}")
    print(f"- Empty Text: {error_cnt['empty_text']}")
    print(f"- Invalid Label: {error_cnt['invalid_label']}")
    print(f"- Duplicate ID: {error_cnt['duplicate_id']}")
    print(f"\nValid Data Sources: {sorted(valid_sources)}")
    print(f"Validation Rate: {len(valid_samples)/len(dataset)*100:.0f}%")

validate_dataset(raw_dataset);
````

### Exercise 2: AI Model Training Statistics Analyzer

You are an AI Engineer responsible for monitoring the performance of a machine learning model during training.

Your model produces training and validation metrics after every epoch.

Your task is to build a training statistics analyzer that processes these metrics and generates a performance report.

**Requirements**

Create a function called `analyze_training()`.

1. Calculate:
   - Average training loss
   - Average validation loss
   - Average accuracy
   - Best training loss
   - Best validation loss
   - Best accuracy

2. Find the epoch with the lowest training loss.

3. Find the epoch with the lowest validation loss.

4. Find the epoch with the highest accuracy.

5. Use list comprehension to extract epochs where:
   - Training loss is below 0.5.
   - Validation loss is above 0.6.
   - Accuracy is above 0.9.

6. Identify potential overfitting.

   Consider a model potentially overfitting when:
   - Training loss decreases.
   - Validation loss increases.

7. Use a while loop to simulate monitoring the training process.

   Stop monitoring when:
   - Validation loss exceeds 0.9.
   - Or all epochs have been processed.

**Expected Output**

```text
===== AI Training Statistics =====

Total Epochs: 8

Average Training Loss: 0.4338
Average Validation Loss: 0.7475
Average Accuracy: 0.8463

Best Training Loss: 0.15 (Epoch 8)

Epochs with Validation Loss > 0.6:

Epochs with Accuracy > 0.9:

Potential Overfitting: True

Monitoring Stopped: Epoch 1
```

````python
# given dataset
training_history = [
    {"epoch": 1, "train_loss": 0.95, "val_loss": 1.10, "accuracy": 0.65},
    {"epoch": 2, "train_loss": 0.72, "val_loss": 0.85, "accuracy": 0.72},
    {"epoch": 3, "train_loss": 0.55, "val_loss": 0.62, "accuracy": 0.81},
    {"epoch": 4, "train_loss": 0.40, "val_loss": 0.48, "accuracy": 0.86},
    {"epoch": 5, "train_loss": 0.30, "val_loss": 0.52, "accuracy": 0.89},
    {"epoch": 6, "train_loss": 0.22, "val_loss": 0.65, "accuracy": 0.93},
    {"epoch": 7, "train_loss": 0.18, "val_loss": 0.82, "accuracy": 0.95},
    {"epoch": 8, "train_loss": 0.15, "val_loss": 0.95, "accuracy": 0.97}
]


def analyze_training(logs):
    cnt = len(logs)
    avg_train_loss = sum(log["train_loss"] for log in logs) / cnt
    avg_val_loss = sum(log["val_loss"] for log in logs) / cnt
    avg_accuracy = sum(log["accuracy"] for log in logs) / cnt
    best_train_loss = min(log["train_loss"] for log in logs)
    best_val_loss = min(log["val_loss"] for log in logs)
    best_accuracy = max(log["accuracy"] for log in logs)

    epoch_best_train_loss = min(
        logs,
        key=lambda x: x["train_loss"]
    )

    epoch_best_val_loss = min(
        logs,
        key=lambda x: x["val_loss"]
    )

    epoch_best_accuracy = max(
        logs,
        key=lambda x: x["accuracy"]
    )

    # extract epochs with conditions
    low_train_loss = [
        log["epoch"] for log in logs
        if log["train_loss"] < 0.5
    ]

    high_val_loss = [
        log["epoch"] for log in logs
        if log["val_loss"] > 0.6
    ]

    high_accuracy = [
        log["epoch"] for log in logs
        if log["accuracy"] > 0.9
    ]

    # check overfitting
    overfitting = False
    for i in range(1, cnt):
        if logs[i]["train_loss"] < logs[i - 1]["train_loss"] and logs[i]["val_loss"] > logs[i - 1]["val_loss"]:
            overfitting = True
            break

    # monitoring
    idx = 0
    reason = "All epochs have been processed"
    while idx < cnt:
        if logs[idx]["val_loss"] > 0.9:
            reason = "Validation loss exceeded 0.9"
            break
        idx += 1

    # print
    print("===== AI Training Statistics =====")

    print(f"Total Epochs: {cnt}")

    print(f"Average Training Loss: {avg_train_loss:.4f}")
    print(f"Average Validation Loss: {avg_val_loss:.4f}")
    print(f"Average Accuracy: {avg_accuracy:.4f}")

    print(f"Best Training Loss: {best_train_loss} (Epoch {epoch_best_train_loss["epoch"]})")
    print(f"Best Validation Loss: {best_val_loss} (Epoch {epoch_best_val_loss["epoch"]})")
    print(f"Best Accuracy: {best_accuracy} (Epoch {epoch_best_accuracy["epoch"]})")

    print(f"Epochs with Training Loss < 0.5: {low_train_loss}")
    print(f"Epochs with Validation Loss > 0.6: {high_val_loss}")
    print(f"Epochs with Accuracy > 0.9: {high_accuracy}")

    print(f"Potential Overfitting: {overfitting}")

    print(f"Monitoring Stopped: Epoch {idx+1}")
    print(f"Reason: {reason}")

analyze_training(training_history)
````

### Exercise 3: NLP Text Processing Pipeline

You are developing a Natural Language Processing (NLP) pipeline for an AI-powered sentiment analysis system.

The system receives customer reviews and must preprocess the text before sending it to a machine learning model.

Your task is to implement a text preprocessing pipeline.


**Requirements**

Create the following functions:

1. `clean_text(text)`
   - Convert text to lowercase.
   - Remove punctuation.
   - Remove unnecessary whitespace.
   - Handle None and empty strings.

2. `tokenize_text(text)`
   - Split text into individual words.
   - Return a list of tokens.

3. `remove_stopwords(tokens)`
   - Remove common stopwords.
   - Use a set for efficient lookup.

4. `calculate_word_frequency(tokens)`
   - Count how frequently each word appears.
   - Return a dictionary.

5. `process_reviews(reviews)`
   - Process all reviews.
   - Separate valid and invalid reviews.
   - Store processed tokens in each valid review.

6. Use list comprehension to extract all unique words.

7. Use dictionary comprehension to create a word frequency summary.

8. Use lambda to sort words by frequency in descending order.

9. Use tuple unpacking to extract the most frequent word and its frequency.

10. Use a generator expression to calculate the total number of processed tokens.

11. Use a decorator to measure and print the execution time of `process_reviews()`.

12. Use exception handling to handle unexpected input types.

**Expected Output**

```text
===== NLP Processing Report =====

Total Reviews: 8
Valid Reviews: 6
Invalid Reviews: 2

Total Processed Tokens: 19

Top 5 Most Frequent Words:

- ai: 4
- product: 2
- bad: 2
- experience: 2
- love: 1

Unique Words: 15

Most Frequent Word: ai
Frequency: 4

Processing Completed Successfully.
```

````python
# given dataset
reviews = [
    {"id": 1, "text": "I LOVE this AI product!", "rating": 5},
    {"id": 2, "text": "The performance is NOT good.", "rating": 2},
    {"id": 3, "text": "Amazing experience with machine learning!", "rating": 5},
    {"id": 4, "text": "   ", "rating": 3},
    {"id": 5, "text": None, "rating": 1},
    {"id": 6, "text": "Python is useful and powerful.", "rating": 4},
    {"id": 7, "text": "AI AI AI is everywhere!", "rating": 5},
    {"id": 8, "text": "Bad product. Very bad experience.", "rating": 1}
]
stopwords = {"is", "the", "and", "a", "this", "with"}

import re
import time
from functools import wraps


def calculate_time(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        start = time.time()
        result = func(*args, **kwargs)
        end = time.time()

        print(f"Processing Time: {end - start:.6f} seconds")
        return result
    return wrapper


def clean_text(text):
    if text is None:
        cleaned_text = ""
    else: 
        cleaned_text = text.lower().strip()
        cleaned_text = re.sub(r'[^a-zA-Z0-9\s]', '', cleaned_text)
    return cleaned_text


def tokenize_text(text):
    return text.split()


def remove_stopwords(tokens):
    return [word for word in tokens if word not in stopwords]


def calculate_word_frequency(tokens):
    word_frequency = {}
    for word in tokens:
        if word in word_frequency:
            word_frequency[word] += 1
        else:
            word_frequency[word] = 1
    return word_frequency

@calculate_time
def process_reviews(reviews):
    required_fields = ["id", "text", "rating"]
    valid_reviews = []
    invalid_reviews = []
    
    for review in reviews:
        try:
            if not isinstance(review, dict):
                invalid_reviews.append(review)
                continue

            id = review.get("id")
            text = review.get("text")
            rating = review.get("rating")

            if (
                not all(field in review for field in required_fields)
                or not isinstance(id, int)
                or not isinstance(text, str)
                or not text.strip()
                or not isinstance(rating, (int, float))
                or not 1 <= rating <= 5
            ):
                invalid_reviews.append(review)
                continue

            cleaned_text = clean_text(text)
            tokens = tokenize_text(cleaned_text)
            tokens = remove_stopwords(tokens)

            processed_review = review.copy()
            processed_review["tokens"] = tokens

            valid_reviews.append(processed_review)

        except (TypeError, ValueError, AttributeError):
            invalid_reviews.append(review)

    return valid_reviews, invalid_reviews


valid_reviews, invalid_reviews = process_reviews(reviews)

all_tokens = [
    token
    for review in valid_reviews
    for token in review['tokens']
]

word_frequency = calculate_word_frequency(all_tokens)

frequency_summary = {
    word: count
    for word, count in word_frequency.items()
}

unique_words = list(set([word for word in all_tokens]))

sorted_words = sorted(
    frequency_summary.items(),
    key=lambda item: item[1],
    reverse=True
)

most_frequent_word, frequency = sorted_words[0]

total_tokens = sum(len(review["tokens"]) for review in valid_reviews)


# print
print("===== NLP Processing Report =====\n")

print(f"Total Reviews: {len(reviews)}")
print(f"Valid Reviews: {len(valid_reviews)}")
print(f"Invalid Reviews: {len(invalid_reviews)}")

print(f"\nTotal Processed Tokens: {total_tokens}")

print("\nTop 5 Most Frequent Words:\n")

for word, count in sorted_words[:5]:
    print(f"- {word}: {count}")


print(f"\nUnique Words: {len(unique_words)}")

print(f"\nMost Frequent Word: {most_frequent_word}")
print(f"Frequency: {frequency}")

print("\nProcessing Completed Successfully.")



````

````python
# Given Experiment Data
experiments = [
    {
        "id": 1,
        "model": "RandomForest",
        "learning_rate": 0.01,
        "accuracy": 0.82,
        "status": "completed"
    },
    {
        "id": 2,
        "model": "NeuralNetwork",
        "learning_rate": 0.001,
        "accuracy": 0.94,
        "status": "completed"
    },
    {
        "id": 3,
        "model": "XGBoost",
        "learning_rate": 0.05,
        "accuracy": 0.89,
        "status": "completed"
    },
    {
        "id": 4,
        "model": "NeuralNetwork",
        "learning_rate": 0.01,
        "accuracy": None,
        "status": "running"
    },
    {
        "id": 5,
        "model": "SVM",
        "learning_rate": 0.1,
        "accuracy": 0.76,
        "status": "failed"
    },
    {
        "id": 6,
        "model": "NeuralNetwork",
        "learning_rate": 0.0005,
        "accuracy": 0.96,
        "status": "completed"
    }
]


class DuplicateIdError(Exception):
    pass

class ExperimentNotFoundError(Exception):
    pass


def log_execution(func):
    def wrapper(*args, **kwargs):
        print(f"Executing {func.__name__}")
        result = func(*args, **kwargs)
        return result
    return wrapper


class ExperimentManager:
    def __init__(self, experiments):
        self.experiments = experiments

    @log_execution
    def add_experiment(self, experiment):
        for exp in self.experiments:
            if exp["id"] == experiment["id"]:
                raise DuplicateIdError(
                    f"Experiment ID {experiment["id"]} already exists."
                )
        self.experiments.append(experiment)

        print("Experiment added successfully")

    @log_execution
    def remove_experiment(self, experiment_id):
        for exp in self.experiments:
            if exp["id"] == experiment_id:
                self.experiments.remove(exp)
                print(f"Experiment {experiment_id} removed successfully.")
                return
            
        raise ExperimentNotFoundError(
            f"Experiment ID {experiment_id} not found."
        )

    @log_execution
    def get_experiment(self, experiment_id):
        for exp in self.experiments:
            if exp["id"] == experiment_id:
                return exp
            
        raise ExperimentNotFoundError(
            f"Experiment ID {experiment_id} not found."
        )

    @log_execution
    def get_completed_experiments(self):
        return [
            exp for exp in self.experiments 
            if exp["status"] == "completed"
        ]

    @log_execution
    def get_best_experiment(self):
        valid_experiments = [
            exp for exp in self.experiments
            if exp["accuracy"] is not None
        ]
        
        if not valid_experiments:
            return None
        
        return max(
            valid_experiments,
            key=lambda exp: exp["accuracy"] 
        )

    @log_execution
    def group_by_model(self):
        models = set(exp["model"] for exp in self.experiments)

        return {
            model: [
                exp for exp in self.experiments
                if exp["model"] == model
            ]
            for model in models
        }

    @log_execution
    def calculate_model_statistics(self):
        models = set(exp["model"] for exp in self.experiments)

        avg_accuracy = {}
        for model in models:
            valid_accuracies = [
                exp["accuracy"] for exp in experiments
                if exp["model"] == model and exp["accuracy"] is not None
            ]
            if valid_accuracies:
                avg_accuracy[model] = sum(valid_accuracies) / len(valid_accuracies)
            else:
                avg_accuracy[model] = None

        status_count = {
            status: sum(
                1 for exp in self.experiments
                if exp["status"] == status
            )
            for status in ["completed", "running", "failed"]
        }

        return avg_accuracy, status_count

    @log_execution
    def rank_experiments(self):
        return sorted(
            self.experiments,
            key=lambda exp: exp["accuracy"],
            reverse=True
        )


    def completed_experiment_generator(self):
        for exp in self.experiments:
            if exp["status"] == "completed":
                yield exp


    def save_report(self, filename):
        with open(filename, "w", encoding="utf-8") as file:
            file.write("===== AI Experiment Manager =====\n\n")
            file.write(f"Total Experiments: {len(self.experiments)}\n\n")
            
            file.write("Experiment Status:\n")
            avg_accuracy, status_count = self.calculate_model_statistics()
            for status, count in status_count.items():
                file.write(f"- {status.capitalize()}: {count}\n")
            
            file.write("\nBest Experiment:\n")
            best = self.get_best_experiment()
            file.write(f"- ID: {best["id"]}\n")
            file.write(f"- Model: {best["model"]}\n")
            file.write(f"- Accuracy: {best["accuracy"]}\n")

            file.write("\nExperiment Ranking:\n\n")
            ranking = self.rank_experiments()
            for i, exp in enumerate(ranking, start=1):
                file.write(f"{i}. {exp["model"] - {exp["accuracy"]}}\n")

            file.write("\nAverage Accuracy by Model:\n\n")
            for model, accuracy in avg_accuracy.item():
                if accuracy is not None:
                    file.write(f"- {model}: {accuracy:.2f}\n")
                else:
                    file.write(f"- {model}: N/A\n")

            print("Experiment Report Saved Successfully.")
````
